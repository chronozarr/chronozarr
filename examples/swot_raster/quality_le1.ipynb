# Local SWOT float32 demo
Two official 100 m SWOT raster dates over the Roanoke region, cropped to the same native-grid window. No resampling, interpolation or cloud upload. The mask retains only wse_qual <= 1 (good or suspect). This tests software fidelity, not elevation quality. The store is about 1 MB.

```python
from pathlib import Path

import chronozarr

repo = Path.cwd()
if not (repo / "src/chronozarr").exists():
    repo = repo.parents[1]
store_path = repo / "data/stores/swot_roanoke/usable-20261002"
store = chronozarr.open_store(store_path)
print(store.times, store.attrs.bands)
```

```python
movie = chronozarr.player(store_path)
movie
```

If your browser asks for local-network access, allow it: the hosted viewer reads this local store. The validity mask excludes source fill/nonfinite values and degraded/bad quality flags; negative retained values are preserved. Genuine observation gaps remain; nothing is filled or smoothed. No exact-zero valid WSE values occur in this crop.

```python
movie.t = 1
movie.click
```

```python
import xarray as xr

ds = xr.open_dataset(store_path, engine="chronozarr")
ds
```

Units are preserved as m in the store, xarray interfaces and COG export. Values and masks match. Do not infer physical quality from the display stretch.

```python
import leafmap.maplibregl as leafmap
from chronozarr import add_chronozarr

from chronozarr.view import serve_store

m = leafmap.Map(style="positron", height="600px",
    add_sidebar=False, add_floating_sidebar=False)
add_chronozarr(m, serve_store(store_path).url, name="SWOT WSE (m)", product="band")
m
```
