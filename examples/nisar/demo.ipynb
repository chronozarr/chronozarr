# Local NISAR cached-extract demo

Two 20 m HH/HV extracts dated 2026-06-22 and 2026-08-21 by filename. Full acquisition metadata was not recovered. This is a software demo, not a calibrated change analysis. Sources remain untouched.

The store retains linear HH/HV and adds explicit `10*log10(linear)` dB bands. Maps use the same −25 to 0 dB grayscale limits on both dates; values outside those limits are clipped only for display. No interpolation or smoothing is applied.

```python
import sys
from pathlib import Path

import chronozarr

repo = Path.cwd()
if not (repo / "src/chronozarr").exists():
    repo = repo.parents[1]
sys.path.insert(0, str(repo / "examples/leafmap"))
store_path = repo / "data/stores/nisar/local-20261002"
store = chronozarr.open_store(store_path)
print(store.times, store.attrs.bands)
```

## HH backscatter

Fixed grayscale: black at −25 dB, white at 0 dB. Use the time slider below the map.

```python
import leafmap.maplibregl as leafmap
from chronozarr_map import add_chronozarr

from chronozarr.view import serve_store

url = serve_store(store_path).url
m = leafmap.Map(style="positron", height="600px", add_sidebar=False, add_floating_sidebar=False)
add_chronozarr(m, url, name="NISAR HH (dB)", product="band", band=0, range=[-25, 0])
m
```

## HV backscatter

The same fixed limits and dates; HV is typically darker.

```python
hv = leafmap.Map(style="positron", height="600px", add_sidebar=False, add_floating_sidebar=False)
add_chronozarr(hv, url, name="NISAR HV (dB)", product="band", band=1, range=[-25, 0])
hv
```
