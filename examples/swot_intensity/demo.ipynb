# SWOT intensity from existing geocoded amplitude
Wax Lake / Atchafalaya: 2024-12-11 and 2025-05-06 on a matching 60 m grid. Original amplitude is preserved; intensity = amplitude² and intensity dB = 10 log10(intensity) are derived **after geocoding**. This is uncalibrated intensity, not sigma0. Nonpositive amplitudes are excluded from the shared mask for logarithmic display. No additional interpolation or smoothing.


```python
from pathlib import Path

import chronozarr

repo = Path.cwd()
if not (repo / "src/chronozarr").exists():
    repo = repo.parents[1]
store_path = repo / "data/stores/swot_intensity/local-20261002"
store = chronozarr.open_store(store_path)
print(store.times, store.attrs.bands)
```

```python
widget = chronozarr.player(store_path, product="band", band="intensity_dB", range=[30, 80])
widget
```

The player defaults to play and a time slider. Set `widget.controls = True` to reveal band selection and display limits.
