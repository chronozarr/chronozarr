# chronozarr on leafmap
Direct static-store pixels on leafmap's MapLibre map. Install leafmap in this notebook's environment, then run the cells.

```python
import leafmap.maplibregl as leafmap

from chronozarr import add_chronozarr
```

```python
m = leafmap.Map(style="positron", height="600px", add_sidebar=False, add_floating_sidebar=False)
add_chronozarr(m)
m
```

The slider selects monthly timesteps. To try the rendered PNG frames instead, create a fresh map and supply the store URL below. RGB values are display colours, not reflectance.

```python
png_map = leafmap.Map(
    style="positron", height="600px", add_sidebar=False, add_floating_sidebar=False
)
add_chronozarr(
    png_map, "https://data.chronozarr.org/ucayali_santa_maria/png-v03", name="PNG frames"
)
png_map
```
